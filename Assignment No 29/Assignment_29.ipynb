### Q29: Import matplotlib.pyplot and plot a histogram of `Severity`.


```python
import matplotlib.pyplot as plt
plt.hist(df['Severity'], bins=10)
plt.xlabel('Severity')
plt.ylabel('Frequency')
plt.title('Histogram of Severity')
plt.show()
```

### Q30: Plot a bar chart showing the top 10 most frequent values in `ID`.


```python
df['ID'].value_counts().head(10).plot(kind='bar')

plt.xlabel('ID')
plt.ylabel('Frequency')
plt.title('Top 10 Most Frequent ID Values')
plt.xticks(rotation=45)
plt.show()
```

### Q31: Plot a scatter plot between `Severity` and `Start_Lat`.


```python
plt.scatter(df['Severity'], df['Start_Lat'])

plt.xlabel('Severity')
plt.ylabel('Start_Lat')
plt.title('Scatter Plot: Severity vs Start_Lat')
plt.show()
```

### Q32: Group by `Source` and plot the sum of `Severity` as a pie chart (top 5 categories).


```python
df.groupby('Source')['Severity'].sum().nlargest(5).plot(kind='pie', autopct='%1.1f%%')

plt.title('Top 5 Sources by Total Severity')
plt.ylabel('')
plt.show()
```

### Q33: Find the row with the maximum `Severity` and display its details.

### Q34: Create a pivot table with `ID` as index and `Severity` as values.


```python
df.pivot_table(index='ID', values='Severity')
```

### Q35: Use `.apply()` to create a new column indicating whether `Severity` is 'High' (above mean) or 'Low' (below mean).


```python
mean_severity = df['Severity'].mean()

df['Severity_Level'] = df['Severity'].apply(
    lambda x: 'High' if x > mean_severity else 'Low'
)

df
```
