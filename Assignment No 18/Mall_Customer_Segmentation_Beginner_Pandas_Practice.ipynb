## About the Dataset

The **Mall_Customers.csv** dataset contains basic details of customers of a shopping mall, along with their yearly income and a spending score given by the mall.

**Source:** Kaggle - Customer Segmentation Tutorial in Python (https://www.kaggle.com/datasets/vjchoudhary7/customer-segmentation-tutorial-in-python).

The dataset contains:

- **200 rows**
- **5 columns**
- **No missing values were found during the initial dataset review**

### Columns in the Dataset

| Column | Description |
|---|---|
| `CustomerID` | Unique ID of the customer |
| `Genre` | Gender of the customer (Male / Female) |
| `Age` | Age of the customer |
| `Annual Income (k$)` | Annual income in thousands of dollars |
| `Spending Score (1-100)` | Score given by the mall based on customer spending behaviour |
